```python
import torch
import torch.nn.functional as F
from ca_sae.sae.ca_sae import ClassAlignedSAE
from lapsum.topk import soft_topk
```

```python
sae = ClassAlignedSAE.from_pretrained("../checkpoints/test/ca_sae_v2/")
matrix_a = soft_topk(sae.class_matrix.cpu(), 4096 * 5 * torch.softmax(sae.budget_vector.cpu(), dim=0).unsqueeze(1), alpha=0.01)
# matrix_a = torch.load("../experiment_results/empirical_matrices/batch_topk_63_train")
```

```python
matrix_a_norm = F.normalize(matrix_a)
sim_a = matrix_a_norm @ matrix_a_norm.T

mean_sim = sim_a[torch.triu(torch.ones_like(sim_a, dtype=torch.bool), diagonal=1)].mean()
mean_sim
```
