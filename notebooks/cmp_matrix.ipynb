```python
import torch
import matplotlib.pyplot as plt
```

```python
matrix_a = torch.load("../experiment_results/empirical_matrices/batch_topk_63")
matrix_b = torch.load("../experiment_results/empirical_matrices/ca_sav2")
```

```python
ks_a = matrix_a.sum(dim=1)
ks_b = matrix_b.sum(dim=1)
```

```python
plt.hist(ks_a.numpy(), bins=20, alpha=0.5, label="Batch", log=True)
plt.hist(ks_b.numpy(), bins=20, alpha=0.5, label="CA", log=True)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
cs_a = matrix_a.sum(dim=0)
cs_b = matrix_b.sum(dim=0)
```

```python
plt.hist(cs_a.numpy(), bins=20, alpha=0.5, label="Batch")
plt.hist(cs_b.numpy(), bins=20, alpha=0.5, label="CA")

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

cutout_region = 100

# First heatmap
im1 = axes[0].imshow(
    matrix_a[:cutout_region, :cutout_region].detach().cpu().numpy(),
    aspect="auto",
    cmap="viridis"
)
fig.colorbar(im1, ax=axes[0], label="Value")
axes[0].set_xlabel("Width (1000)")
axes[0].set_ylabel("Height (4096)")
axes[0].set_title("Matrix A")

# Second heatmap
im2 = axes[1].imshow(
    matrix_b[:cutout_region, :cutout_region].detach().cpu().numpy(),
    aspect="auto",
    cmap="viridis"
)
fig.colorbar(im2, ax=axes[1], label="Value")
axes[1].set_xlabel("Width (1000)")
axes[1].set_ylabel("Height (4096)")
axes[1].set_title("Matrix B")

plt.tight_layout()
plt.show()
```
