```python
import torch
from lapsum.topk import soft_topk
```

```python
logits = torch.randn((10, 4096))
m_lambda = torch.randn(4096, 1000)
alpha = torch.tensor(0.04)
k_hat = torch.rand(10) * 40
p = soft_topk(logits, k_hat.unsqueeze(-1), alpha)
pi = p / k_hat.unsqueeze(-1)
labels = torch.randint(0, 1000, size=(10,))

k = 4096 * 5 * torch.softmax(torch.randn((4096,)), dim=0)

M = soft_topk(m_lambda, k.unsqueeze(-1), alpha, dim=1)
M_class = M[:,  labels].T

print((pi * M_class).sum(dim=1))

M_cl = M[:, labels]
(pi @ M_cl).diag()
```
