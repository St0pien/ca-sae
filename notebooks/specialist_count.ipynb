```python
import json
import matplotlib.pyplot as plt
```

```python
with open("../experiment_results/m_honesty/batch_topk_63_empirical_honesty_estimated.json") as f:
    one = json.load(f)

with open("../experiment_results/m_honesty/ca_sa_v2.json") as f:
    two = json.load(f)
```

```python
first_distribution = [len(f['claimed_classes']) for f in one['features']]
second_distribution = [len(f['claimed_classes']) for f in two['features']]
```

```python
plt.hist(first_distribution, bins=30, alpha=0.5, log=True, label="Batch")
plt.hist(second_distribution, bins=30, alpha=0.5, log=True, label="CA SAE")
plt.legend()
```
