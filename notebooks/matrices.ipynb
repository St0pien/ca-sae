```python
import torch
import matplotlib.pyplot as plt
from lapsum.topk import soft_topk
```

```python
idx = 1000
vector = torch.load(f"../matrices/vector_{idx}.pt").detach()

k = 5 * 4096 * torch.softmax(vector, dim=0)
heatmap  = soft_topk(torch.load(f"../matrices/matrix_{idx}.pt"), k.unsqueeze(-1), 0.001)
```

```python

# heatmap: torch.Tensor with shape (4096, 1000)

plt.figure(figsize=(12, 8))
plt.imshow(
    heatmap[:100, :100].detach().cpu().numpy(),
    aspect="auto",
    cmap="viridis"
)
plt.colorbar(label="Value")
plt.xlabel("Width (1000)")
plt.ylabel("Height (4096)")
plt.title("Heatmap")
plt.show()
```
